**B. Kueri Bertingkat (Read)**

Cell 1: Menampilkan menu kategori Kopi dengan tingkat kepahitan di atas 3

```python
print("Menu Kopi dengan tingkat kepahitan > 3:")

for menu in koleksi_menu.find({
    "kategori": "Kopi",
    "tingkat_kepahitan": {"$gt": 3}
}, {"_id": 0}):
    print(menu)
```

Cell 2: Menampilkan menu dengan harga antara Rp15.000 sampai Rp30.000

```python
print("Menu dengan harga Rp15.000 - Rp30.000:")

for menu in koleksi_menu.find({
    "harga": {
        "$gte": 15000,
        "$lte": 30000
    }
}, {"_id": 0}):
    print(menu)
```

Cell 3: Menampilkan menu yang tidak memiliki field pedas_level

```python
print("Menu yang tidak memiliki field pedas_level:")

for menu in koleksi_menu.find({
    "pedas_level": {"$exists": False}
}, {"_id": 0}):
    print(menu)
```

**C. Pembaruan Data (Update)**

Cell 1: Menambahkan Status ke Seluruh Menu

```python
hasil_update = koleksi_menu.update_many(
    {},
    {
        "$set": {
            "status": "Tersedia"
        }
    }
)

print("Jumlah dokumen yang diperbarui:", hasil_update.modified_count)
```

Cell 2: Mengubah Status Dua Menu Menjadi Habis

```python
koleksi_menu.update_one(
    {"nama_menu": "Espresso"},
    {"$set": {"status": "Habis"}}
)

koleksi_menu.update_one(
    {"nama_menu": "Nasi Goreng"},
    {"$set": {"status": "Habis"}}
)

print("Status Espresso dan Nasi Goreng berhasil diperbarui!")
```

Cell 3: Menampilkan Hasil Update

```python
print("Daftar Menu Setelah Update:")

for menu in koleksi_menu.find({}, {"_id": 0}):
    print(
        menu["nama_menu"],
        "-",
        menu["kategori"],
        "-",
        menu["status"]
    )
```

**D. Integrasi ke Spark**

Cell 1: Membuat SparkSession dan Mengambil Data MongoDB

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType, StructField,
    StringType, IntegerType
)
from pyspark.sql.functions import avg, count
import pandas as pd

# Membuat SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

# Mengambil seluruh data dari MongoDB
dokumen_menu = list(koleksi_menu.find({}))

# Mengubah data menjadi Pandas DataFrame
df_pandas = pd.DataFrame(dokumen_menu)

# Menghapus kolom _id MongoDB
df_pandas = df_pandas.drop(columns=["_id"])

print("Data berhasil diambil dari MongoDB")
print(df_pandas)
```

Cell 2: Menyesuaikan Struktur Data

```python
# Membuat skema Spark secara eksplisit
schema = StructType([
    StructField("nama_menu", StringType(), True),
    StructField("harga", IntegerType(), True),
    StructField("kategori", StringType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("stok", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("status", StringType(), True)
])

# Menyesuaikan data agar nilai kosong menjadi None
data_spark = []

for _, row in df_pandas.iterrows():
    data_spark.append({
        "nama_menu": row["nama_menu"],
        "harga": int(row["harga"]),
        "kategori": row["kategori"],
        "tingkat_kepahitan": (
            int(row["tingkat_kepahitan"])
            if pd.notna(row.get("tingkat_kepahitan"))
            else None
        ),
        "stok": int(row["stok"]),
        "pedas_level": (
            int(row["pedas_level"])
            if pd.notna(row.get("pedas_level"))
            else None
        ),
        "status": row["status"]
    })

# Membuat Spark DataFrame
df_spark = spark.createDataFrame(data_spark, schema=schema)

print("Spark DataFrame berhasil dibuat!")
```

Cell 3: Menampilkan Skema Data

```python
df_spark.printSchema()
```

Cell 4: Menampilkan Seluruh Data Spark DataFrame

```python
df_spark.show(truncate=False)
```

Cell 5: Menghitung Rata-Rata Harga Berdasarkan Kategori

```python
hasil_analisis = df_spark.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
)

hasil_analisis.show()
```

Cell 6: Menghentikan SparkSession

```python
spark.stop()

print("SparkSession berhasil dihentikan.")
```
